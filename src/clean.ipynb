```python
import pandas as pd
```

```python
df = pd.read_csv("C:\\Users\\ELGRM\\OneDrive\\Desktop\\premier league predictor\\data\\raw\\raw_data.csv", dtype={"Season_code": str})
```

```python
df
```

```python
df["HomeTeam"]
```

```python
name_new ={
    "Man Utd": "Manchester United",
    "Man City": "Manchester City",
    "Man United": "Manchester United",
    "Tottenham": "Tottenham Hotspur",
}
df["HomeTeam"] = df["HomeTeam"].replace(name_new)
df["AwayTeam"] = df["AwayTeam"].replace(name_new)
```

```python
df  = df.drop_duplicates()
```

```python
df = df.dropna(subset=["HomeTeam", "AwayTeam" , "Date" ,"FTR"])
```

```python
df
```

```python
df.columns
```

```python
core_cols = [
    # Match info
    "Season_code", "Date", "HomeTeam", "AwayTeam",
    # Full-time & Half-time Targets
    "FTHG", "FTAG", "FTR", "HTHG", "HTAG", "HTR",
    # Match stats (for form & feature calculation)
    "HS", "AS", "HST", "AST", "HC", "AC", "HF", "AF", "HY", "AY", "HR", "AR", "Referee",
    # Benchmark Odds (Bet365 & Market Average)
    "B365H", "B365D", "B365A"
]

# Keep only the columns that exist in the raw dataframe
df_clean = df[[col for col in core_cols if col in df.columns]].copy()
```

```python
df_clean
```

```python
df_clean.info()
```

```python
df_clean["Date"] = pd.to_datetime(df_clean["Date"], dayfirst=True, errors="coerce")
```

```python
df_clean = df_clean.sort_values("Date").reset_index(drop=True)
```

```python
df_clean
```

```python

from pathlib import Path


Path("data/processed").mkdir(parents=True, exist_ok=True)
df_clean.to_csv("data/processed/matches_clean.csv", index=False)
```
